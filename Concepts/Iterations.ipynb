```python
animals = ['cat', 'dog', 'pig']
for index in range(len(animals)):
    print(index,animals[index])
```

```python
for index,value in enumerate(animals):
    print(index,value)
```

```python
animals = ['cat', 'dog', 'bird']
for index,value in enumerate(animals):
    if value == 'dog':
        break
    print(index,value)
```

```python
animals = ['cat', 'dog', 'bird']
for index,value in enumerate(animals):
    if value == 'dog':
        continue
    print(index,value)
```

```python
my_dict = {"name":"shalini", "cash": 5.45}
for key in my_dict:
    print(key)
```

```python
for value in my_dict.values():
    print(value)
```

```python
for key,value in my_dict.items():
    print(key,value)
```

```python
names = ['cindy', 'paul', 'michael', 'ringo']
names2 = []
for name in names:
    if len(name) == 4:
        names2.append(name.title())
names2
```

```python
names2 = [name.title() for name in names if len(name) == 4]
```

```python
names2
```

```python
#Dict Comprehensions
```

```python
types = { 'name': str, 'age': int, 'address': str}
```

```python
new_names = {}
for t in types:
    new_names[t] = t.title()
new_names
```

```python
new_names = { t:t.title() for t in types}
```

```python
new_names
```
